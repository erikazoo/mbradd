Data Exploration Begins here

Most common provider

```python
#Provider types with the most unique providers
top_unique_providers = (df.groupby("Rndrng_Prvdr_Type")["Rndrng_NPI"].nunique().sort_values(ascending=False).head(15))
top_unique_providers
```

```python
top_unique_providers.plot(
    kind="barh",
    figsize=(8, 5)
)

plt.xlabel("Number of Records")
plt.ylabel("Provider Type")
plt.title("Top 10 Provider Unique Types")
plt.tight_layout()
plt.show()
```

```python
#radiologists were top1 in service billing chart, nurses in unique type, now checking their contribution to service records
records_by_type = (df.groupby("Rndrng_Prvdr_Type").size())
providers_by_type = (df.groupby("Rndrng_Prvdr_Type")["Rndrng_NPI"].nunique())
records_per_provider = (records_by_type / providers_by_type).sort_values(ascending=False)
records_per_provider.head(15)
```

Most common medical service

```python
top_services = (df["HCPCS_Cd"].value_counts().head(15))
top_services
```

```python
top_services.sort_values().plot(kind="barh", figsize=(10, 6))
plt.xlabel("Number of Provider-Service Records")
plt.ylabel("Service")
plt.title("Most Frequently Appearing Medicare Services")
plt.tight_layout()
plt.show()
```

```python
top_services_volume = (df.groupby(["HCPCS_Cd"])["Tot_Srvcs"].sum().sort_values(ascending=False).head(15))
top_services_volume
```

```python
top_services_volume.sort_values().plot(kind="barh", figsize=(10, 6))
plt.xlabel("Number of Provider-Service Records")
plt.ylabel("Service")
plt.title("Top Services Volume")
plt.tight_layout()
plt.show()
```

```python
service_summary = (df.groupby(["HCPCS_Cd", "HCPCS_Desc"]).agg(record_count=("Rndrng_NPI", "size"), unique_providers=("Rndrng_NPI", "nunique"), total_services=("Tot_Srvcs", "sum")).reset_index())
service_summary["services_per_record"] = (service_summary["total_services"] / service_summary["record_count"])
service_summary.sort_values("record_count", ascending=False).head(10)
```

service per patient

```python
df["services_per_patient"] = (df["Tot_Srvcs"] / df["Tot_Benes"])
```

```python
df["services_per_patient"].describe(percentiles=[0.05, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95, 0.99])
```

```python
top_types = (df["Rndrng_Prvdr_Type"].value_counts().head(10).index)

specialty_medians = (df[df["Rndrng_Prvdr_Type"].isin(top_types)].groupby("Rndrng_Prvdr_Type")["services_per_patient"].median().sort_values())

specialty_medians
```

```python
specialty_medians.plot(kind="barh", figsize=(9,6))

plt.xlabel("Median Services per Patient")
plt.ylabel("Provider Type")
plt.title("Typical Service Intensity by Provider Type")
plt.tight_layout()
plt.show()
```

what specialities work mostly in faacility setting and vice versa

```python
place_by_speciality = pd.crosstab(df["Rndrng_Prvdr_Type"], df["Place_Of_Srvc"], normalize="index") * 100
```

```python
place_by_speciality.loc[top_types]
```

drug records

```python
drug_df = df[df["HCPCS_Drug_Ind"] == "Y"]

drug_specialties = (drug_df["Rndrng_Prvdr_Type"].value_counts().head(15))

drug_specialties
```

```python
drug_specialties.sort_values().plot(kind="barh",figsize=(9,6))

plt.xlabel("Drug-Related Records")
plt.ylabel("Provider Type")
plt.title("Provider Types with the Most Part B Drug Records")
plt.tight_layout()
plt.show()
```

billing behavior

```python
df["charge_ratio"] = (df["Avg_Sbmtd_Chrg"] / df["Avg_Mdcr_Alowd_Amt"].replace(0, np.nan))
```

```python
df["charge_ratio"].describe(percentiles=[0.50, 0.75, 0.90, 0.95, 0.99])
```

```python
charge_by_specialty = (df[df["Rndrng_Prvdr_Type"].isin(top_types)].groupby("Rndrng_Prvdr_Type")["charge_ratio"].median().sort_values())

charge_by_specialty
```

```python
charge_by_specialty.plot(kind="barh",figsize=(9,6))

plt.xlabel("Median Submitted Charge / Allowed Amount")
plt.ylabel("Provider Type")
plt.title("Typical Charge-to-Allowed Ratio by Provider Type")
plt.tight_layout()
plt.show()
```
