```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from scipy.stats import spearmanr

import shap
```

```python
file = "data/Medicare_Physician_Other_Practitioners_by_Provider_and_Service_2024.csv"

peer_cols = [
    "Rndrng_Prvdr_Type",
    "HCPCS_Cd",
    "Place_Of_Srvc"
]

raw_numeric_cols = [
    "Tot_Benes",
    "Tot_Srvcs",
    "Tot_Bene_Day_Srvcs",
    "Avg_Sbmtd_Chrg",
    "Avg_Mdcr_Alowd_Amt",
    "Avg_Mdcr_Pymt_Amt",
    "Avg_Mdcr_Stdzd_Amt"
]

use_cols = (
    ["Rndrng_NPI"]
    + peer_cols
    + raw_numeric_cols
)

df = pd.read_csv(
    file,
    usecols=use_cols,
    dtype={
        "HCPCS_Cd": "string"
    }
)

print(df.shape)
```

```python
df["services_per_bene"] = (
    df["Tot_Srvcs"] /
    df["Tot_Benes"].replace(0, np.nan)
)

df["services_per_bene_day"] = (
    df["Tot_Srvcs"] /
    df["Tot_Bene_Day_Srvcs"].replace(0, np.nan)
)

df["charge_to_allowed"] = (
    df["Avg_Sbmtd_Chrg"] /
    df["Avg_Mdcr_Alowd_Amt"].replace(0, np.nan)
)

df["payment_to_allowed"] = (
    df["Avg_Mdcr_Pymt_Amt"] /
    df["Avg_Mdcr_Alowd_Amt"].replace(0, np.nan)
)

df.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)
```

```python
candidate_features = [
    "Tot_Benes",
    "Tot_Srvcs",
    "Tot_Bene_Day_Srvcs",
    "services_per_bene",
    "services_per_bene_day",
    "Avg_Sbmtd_Chrg",
    "Avg_Mdcr_Alowd_Amt",
    "Avg_Mdcr_Pymt_Amt",
    "Avg_Mdcr_Stdzd_Amt",
    "charge_to_allowed",
    "payment_to_allowed"
]

current_features = [
    "Tot_Benes",
    "services_per_bene",
    "services_per_bene_day",
    "charge_to_allowed",
    "payment_to_allowed",
    "Avg_Sbmtd_Chrg"
]
```

```python
required_cols = (
    peer_cols
    + candidate_features
)

df_model = (
    df
    .dropna(subset=required_cols)
    .copy()
)

print("Original rows:", len(df))
print("Complete candidate-feature rows:", len(df_model))
```

```python
negative_counts = (
    df_model[candidate_features] < 0
).sum()

negative_counts
```

```python
peer_summary = (
    df_model
    .groupby(peer_cols)
    .size()
    .reset_index(name="group_size")
)

peer_summary["group_size"].describe()
```

```python
MIN_TEST_PEER_SIZE = 500
N_TEST_PEERS = 50

eligible_peers = peer_summary[
    peer_summary["group_size"] >= MIN_TEST_PEER_SIZE
].copy()

test_peers = eligible_peers.sample(
    n=min(
        N_TEST_PEERS,
        len(eligible_peers)
    ),
    random_state=42
)

test_peers.head()
```

```python
selected_keys = set(
    map(
        tuple,
        test_peers[peer_cols].to_numpy()
    )
)

peer_data = {}

for key, group in df_model.groupby(
    peer_cols,
    sort=False
):

    if key in selected_keys:
        peer_data[key] = group.copy()

print("Peer groups collected:", len(peer_data))
```

```python
variation_results = []

for peer_key, group in peer_data.items():

    for feature in candidate_features:

        values = np.log1p(
            group[feature].to_numpy()
        )

        q25 = np.quantile(values, 0.25)
        q75 = np.quantile(values, 0.75)

        variation_results.append({
            "peer": peer_key,
            "feature": feature,
            "std_log": np.std(values),
            "iqr_log": q75 - q25,
            "n_unique": group[feature].nunique()
        })

variation_results = pd.DataFrame(
    variation_results
)
```

```python
variation_summary = (
    variation_results
    .groupby("feature")
    .agg(
        median_peer_std=(
            "std_log",
            "median"
        ),
        median_peer_iqr=(
            "iqr_log",
            "median"
        ),
        median_unique_values=(
            "n_unique",
            "median"
        ),
        pct_constant_peers=(
            "n_unique",
            lambda x: (
                x <= 1
            ).mean() * 100
        )
    )
    .sort_values(
        "median_peer_std",
        ascending=False
    )
)

variation_summary
```

```python
peer_corr_matrices = []

for peer_key, group in peer_data.items():

    corr = (
        group[candidate_features]
        .corr(method="spearman")
        .to_numpy()
    )

    peer_corr_matrices.append(corr)

corr_stack = np.stack(
    peer_corr_matrices
)
```

```python
median_spearman = pd.DataFrame(
    np.nanmedian(
        corr_stack,
        axis=0
    ),
    index=candidate_features,
    columns=candidate_features
)

median_spearman.round(2)
```

```python
median_abs_spearman = pd.DataFrame(
    np.nanmedian(
        np.abs(corr_stack),
        axis=0
    ),
    index=candidate_features,
    columns=candidate_features
)

median_abs_spearman.round(2)
```

```python
fig, ax = plt.subplots(
    figsize=(10, 8)
)

im = ax.imshow(
    median_spearman.values,
    vmin=-1,
    vmax=1
)

ax.set_xticks(
    range(len(candidate_features))
)

ax.set_yticks(
    range(len(candidate_features))
)

ax.set_xticklabels(
    candidate_features,
    rotation=45,
    ha="right"
)

ax.set_yticklabels(
    candidate_features
)

for i in range(len(candidate_features)):
    for j in range(len(candidate_features)):

        ax.text(
            j,
            i,
            f"{median_spearman.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=8
        )

plt.title(
    "Median Within-Peer Spearman Correlation"
)

plt.colorbar(im)

plt.tight_layout()
plt.show()
```

```python
REDUNDANCY_FLAG = 0.85

redundant_pairs = []

for i in range(len(candidate_features)):

    for j in range(i + 1, len(candidate_features)):

        rho = median_abs_spearman.iloc[i, j]

        if rho >= REDUNDANCY_FLAG:

            redundant_pairs.append({
                "feature_1":
                    candidate_features[i],

                "feature_2":
                    candidate_features[j],

                "median_abs_spearman":
                    rho
            })

redundant_pairs = pd.DataFrame(
    redundant_pairs
).sort_values(
    "median_abs_spearman",
    ascending=False
)

redundant_pairs
```

```python
N_ESTIMATORS = 200
MODEL_SEED = 42


def fit_iforest(
    train_df,
    features,
    seed=MODEL_SEED
):

    X_train = np.log1p(
        train_df[features].copy()
    )

    model = IsolationForest(
        n_estimators=N_ESTIMATORS,
        max_samples="auto",
        contamination="auto",
        random_state=seed,
        n_jobs=-1
    )

    model.fit(X_train)

    return model
```

```python
def score_iforest(
    model,
    data,
    features
):

    X = np.log1p(
        data[features].copy()
    )

    scores = -model.score_samples(X)

    return pd.Series(
        scores,
        index=data.index
    )
```

```python
TRAIN_FRAC = 0.70
SPLIT_SEED = 123


def split_peer(group):

    train = group.sample(
        frac=TRAIN_FRAC,
        random_state=SPLIT_SEED
    )

    evaluation = group.drop(
        index=train.index
    )

    return train, evaluation
```

```python
SHAP_PEERS = 10
SHAP_BACKGROUND_SIZE = 20
SHAP_EXPLAIN_SIZE = 40
SHAP_PERMUTATIONS = 5

shap_results = []

shap_peer_items = list(
    peer_data.items()
)[:SHAP_PEERS]
```

```python
for peer_number, (peer_key, group) in enumerate(
    shap_peer_items,
    start=1
):

    train, evaluation = split_peer(group)

    model = fit_iforest(
        train,
        candidate_features
    )

    X_train = np.log1p(
        train[candidate_features]
    )

    X_eval = np.log1p(
        evaluation[candidate_features]
    )

    background = X_train.sample(
        n=min(
            SHAP_BACKGROUND_SIZE,
            len(X_train)
        ),
        random_state=42
    )

    explain_data = X_eval.sample(
        n=min(
            SHAP_EXPLAIN_SIZE,
            len(X_eval)
        ),
        random_state=42
    )

    def anomaly_function(x):

        x_df = pd.DataFrame(
            x,
            columns=candidate_features
        )

        return -model.score_samples(
            x_df
        )

    explainer = shap.Explainer(
        anomaly_function,
        background,
        algorithm="permutation",
        feature_names=candidate_features,
        seed=42
    )

    max_evals = (
        SHAP_PERMUTATIONS
        * 2
        * len(candidate_features)
        + 1
    )

    shap_values = explainer(
        explain_data,
        max_evals=max_evals
    )

    mean_abs_shap = (
        np.abs(shap_values.values)
        .mean(axis=0)
    )

    total_importance = (
        mean_abs_shap.sum()
    )

    for feature, importance in zip(
        candidate_features,
        mean_abs_shap
    ):

        shap_results.append({
            "peer": peer_key,
            "feature": feature,
            "mean_abs_shap": importance,
            "importance_share":
                importance /
                total_importance
        })

    print(
        f"Finished SHAP peer "
        f"{peer_number}/{len(shap_peer_items)}"
    )
```

```python
shap_results = pd.DataFrame(
    shap_results
)

shap_summary = (
    shap_results
    .groupby("feature")
    .agg(
        mean_importance_share=(
            "importance_share",
            "mean"
        ),
        median_importance_share=(
            "importance_share",
            "median"
        ),
        std_importance_share=(
            "importance_share",
            "std"
        )
    )
    .sort_values(
        "mean_importance_share",
        ascending=False
    )
)

shap_summary
```

```python
plot_shap = (
    shap_summary
    .sort_values(
        "mean_importance_share"
    )
)

plt.figure(figsize=(8, 6))

plt.barh(
    plot_shap.index,
    plot_shap[
        "mean_importance_share"
    ]
)

plt.xlabel(
    "Mean Share of Absolute SHAP Importance"
)

plt.ylabel("Feature")

plt.title(
    "Isolation Forest Feature Influence"
)

plt.tight_layout()
plt.show()
```

```python
def top_k_jaccard(
    scores_a,
    scores_b,
    top_pct=0.05
):

    n_top = max(
        1,
        int(
            len(scores_a)
            * top_pct
        )
    )

    top_a = set(
        scores_a.nlargest(
            n_top
        ).index
    )

    top_b = set(
        scores_b.nlargest(
            n_top
        ).index
    )

    intersection = len(
        top_a & top_b
    )

    union = len(
        top_a | top_b
    )

    return intersection / union
```

```python
ablation_results = []

for peer_number, (peer_key, group) in enumerate(
    peer_data.items(),
    start=1
):

    train, evaluation = split_peer(
        group
    )

    # Full candidate model
    full_model = fit_iforest(
        train,
        candidate_features
    )

    full_scores = score_iforest(
        full_model,
        evaluation,
        candidate_features
    )

    # Remove one feature at a time
    for removed_feature in candidate_features:

        reduced_features = [
            feature
            for feature in candidate_features
            if feature != removed_feature
        ]

        reduced_model = fit_iforest(
            train,
            reduced_features
        )

        reduced_scores = score_iforest(
            reduced_model,
            evaluation,
            reduced_features
        )

        rho, _ = spearmanr(
            full_scores,
            reduced_scores
        )

        jaccard_5 = top_k_jaccard(
            full_scores,
            reduced_scores,
            top_pct=0.05
        )

        ablation_results.append({
            "peer": peer_key,
            "removed_feature":
                removed_feature,
            "spearman_vs_full":
                rho,
            "top5_jaccard":
                jaccard_5
        })

    print(
        f"Finished peer "
        f"{peer_number}/{len(peer_data)}"
    )
```

```python
ablation_results = pd.DataFrame(
    ablation_results
)

ablation_summary = (
    ablation_results
    .groupby("removed_feature")
    .agg(
        mean_spearman=(
            "spearman_vs_full",
            "mean"
        ),
        median_spearman=(
            "spearman_vs_full",
            "median"
        ),
        std_spearman=(
            "spearman_vs_full",
            "std"
        ),
        mean_top5_jaccard=(
            "top5_jaccard",
            "mean"
        ),
        median_top5_jaccard=(
            "top5_jaccard",
            "median"
        )
    )
    .sort_values(
        "mean_spearman"
    )
)

ablation_summary
```
